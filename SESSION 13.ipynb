```python
#Reverse a String – Recursive Function
def reverse_string(s):
    if s == "":
        return s

    return reverse_string(s[1:]) + s[0]


result = reverse_string("hello")

print(result)
```

```python
#Calculate Total Playlist Duration – Recursion
def sum_playlist_durations(durations):
    if len(durations) == 0:
        return 0

    return durations[0] + sum_playlist_durations(durations[1:])


durations = [180, 240, 200, 300]

total = sum_playlist_durations(durations)

print("Total playlist duration:", total, "seconds")
```

```python
#Local Scope vs Global Scope
count = 10

def update_count():
    count = 5
    print("Inside:", count)

update_count()

print("Outside:", count)
```

Task 4

```python
def count_likes(post):
    if post == {}:
        return 0

    return post["likes"] + count_likes(post.get("reply", {}))


post = {
    "likes": 100,
    "reply": {
        "likes": 20,
        "reply": {
            "likes": 10
        }
    }
}

print("Total likes:", count_likes(post))
```

```python
#Lifetime of Local and Global Variables
app_status = "Offline"

print("Before function call:")
print("App Status:", app_status)


def update_status():
    user_status = "Online"
    global app_status

    app_status = "Online"

    print("Inside function:")
    print("User Status:", user_status)
    print("App Status:", app_status)


update_status()

print("After function call:")
print("App Status:", app_status)
```
