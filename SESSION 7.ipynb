```python
# nfluencer Classification – IF ELIF ELSE
followers = int(input("Enter number of followers: "))

if followers < 10000:
    print("Micro Influencer")

elif followers <= 100000:
    print("Rising Star")

else:
    print("Celebrity")
```

```python
# Zomato Free Delivery – IF ELIF ELSE
total = int(input("Enter your order total: "))

if total > 299:
    print("Apply Free Delivery")

elif total >= 200:
    print("Add more items for free delivery")

else:
    print("Delivery charges apply")
```

```python
#Flipkart Cashback – NESTED IF
cart_value = int(input("Enter cart value: "))
payment = input("Enter payment method (UPI, Card, Cash): ")

if cart_value > 1000:

    if payment == "UPI":
        print("Eligible for 10% cashback")

    else:
        print("Eligible for 5% cashback")

else:
    print("No cashback")
```
