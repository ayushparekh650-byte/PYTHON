Task 1

```python
#Create a dictionary of Spotify playlist prices

playlist_prices = {
    "Chill Vibes": 99,
    "Party Hits": 149,
    "Workout Mix": 199,
    "Love Songs": 129,
    "Lo-Fi Beats": 79
}

print(playlist_prices)
```

Task 2

```python
# Update playlist price using a function

def update_playlist_price(playlist, new_price):
    playlist_prices[playlist] = new_price

playlist_prices = {
    "Chill Vibes": 99,
    "Party Hits": 149,
    "Workout Mix": 199,
    "Love Songs": 129,
    "Lo-Fi Beats": 79
}

update_playlist_price("Party Hits", 199)

print(playlist_prices)
```

Task 3

```python
#Remove a playlist using del

playlist_prices = {
    "Chill Vibes": 99,
    "Party Hits": 149,
    "Workout Mix": 199,
    "Love Songs": 129,
    "Lo-Fi Beats": 79
}

del playlist_prices["Love Songs"]

print(playlist_prices)
```

```python
#Zomato and Swiggy restaurants using sets

set1 = {"Dominos", "McDonalds", "Subway", "Pizza Hut"}

set2 = {"Subway", "Pizza Hut", "KFC", "Burger King"}

# Union: All unique restaurants from both sets
print("Union:", set1.union(set2))

# Intersection: Common restaurants in both sets
print("Intersection:", set1.intersection(set2))
```
