Task 1

```python
#Calculate Final Price – Function with Arguments & Return
def calculate_final_price(price, discount_rate):
    discount = price * discount_rate
    final_price = price - discount
    return final_price

result = calculate_final_price(1200, 0.15)

print("Final price:", result)
```

```python
#Calculate Cashback – Default Arguments & Return
def calculate_cashback(amount, cashback_rate=0.05):
    cashback = amount * cashback_rate
    return cashback

zomato = calculate_cashback(500)

flipkart = calculate_cashback(2000, 0.07)

print("Zomato cashback:", zomato)
print("Flipkart cashback:", flipkart)
```
