Task 1

```python
#Lambda Function – Add 18% GST
gst = lambda price: price + (price * 0.18)

print(gst(100))
print(gst(250))
print(gst(500))
```

Task 2

```python
#Clean Spotify Song Titles – map() + lambda
songs = [' shape OF you ', '  believer', 'KESARIYA ', ' blinding LIGHTS ']

clean_songs = list(map(lambda song: song.strip().title(), songs))

print(clean_songs)
```

Task 3

```python
#Extract Products Starting with S – filter() + lambda
products = ['Samsung TV', 'Shoes', 'iPhone', 'Smart Watch',
            'Laptop', 'Speaker', 'Mouse']

result = list(filter(lambda name: name.lower().startswith('s'), products))

print(result)
```

```python
#Calculate Total Zomato Bill – reduce()
from functools import reduce

orders = [120, 340, 560, 80]

total = reduce(lambda a, b: a + b, orders)

print("Total bill:", total)
```

Task 5

```python
# Use map(), filter() and reduce() Together
from functools import reduce

numbers = [40, 60, 80, 120]

# Step 1: Double each number
doubled = list(map(lambda x: x * 2, numbers))

# Step 2: Keep numbers greater than 100
filtered = list(filter(lambda x: x > 100, doubled))

# Step 3: Calculate the total
total = reduce(lambda a, b: a + b, filtered)

print("Original numbers:", numbers)
print("Doubled numbers:", doubled)
print("Filtered numbers:", filtered)
print("Final sum:", total)
```
