```python
#Difference between append() and extend():

a = [1, 2]
a.append([3, 4])
print(a)

a = [1, 2]
a.extend([3, 4])
print(a)
```

task 4

```python
# 3. Zomato order using pop()

def remove_last_item(order_list):
    removed_item = order_list.pop()
    return removed_item

order_list = ['pizza', 'burger', 'fries']

result = remove_last_item(order_list)

print("Removed item:", result)
print("Updated order:", order_list)
```

Task 5

```python
#Instagram filters using a tuple

insta_filters = ('Clarendon', 'Juno', 'Lark', 'Gingham')

print("Original tuple:", insta_filters)

try:
    insta_filters[1] = 'Valencia'

except TypeError as e:
    print("Error:", e)

# Tuples are immutable, so their elements cannot be changed
# using direct assignment after the tuple is created.
```
